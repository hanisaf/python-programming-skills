### 1.1 Anatomy of a Python program

Here is a short program. It splits the numbers 0 to 9 into two groups: those below a midpoint and those at or above it.

````python
# set the midpoint
midpoint = 5

# make two empty lists
lower = []
upper = []

# split the numbers into lower and upper
for i in range(10):
    if i < midpoint:
        lower.append(i)
    else:
        upper.append(i)

print("lower:", lower)
print("upper:", upper)
````

Even this small program shows the main rules of Python syntax:

- **Comments start with `#`.** Python ignores everything after `#` on a line. Use comments to explain *why* the code does something.
- **The end of a line ends a statement.** There is no semicolon at the end of `midpoint = 5`.
- **Indentation matters.** The lines indented under `for ...:` form the body of the loop; the lines indented under `if ...:` run only when the condition is true. A colon `:` at the end of a line introduces an indented block. Use four spaces per level.
- **Parentheses group or call.** In `2 * (3 + 4)` they group an operation; in `print("lower:", lower)` they *call* a function and pass it arguments.

````python
2 * (3 + 4)
````

### 1.2 Creating variables

A **variable** is a name that refers to a value. Create one with the assignment operator `=`: the name goes on the left, the value on the right.

````python
price = 19.99
quantity = 3
revenue = price * quantity
revenue
````

The `=` sign means *assign*, not *equals*. The statement `quantity = quantity + 1` would be nonsense in algebra, but in Python it means "compute `quantity + 1`, then make `quantity` refer to the result."

````python
quantity = quantity + 1
quantity
````

**Naming rules.** Names may contain letters, digits, and underscores, but cannot start with a digit or be a Python keyword such as `for`, `if`, or `class`. Names are case sensitive: `Revenue` and `revenue` are different variables.

**Naming conventions.** Python programmers write variable names in lowercase with underscores between words (`total_revenue`, `seconds_per_hour`). Choose names that describe the content: `unit_price` is better than `p` or `x1`.

````python
total_revenue = 59.97   # good: descriptive, snake_case
x = 59.97               # legal, but tells the reader nothing
````

### 1.3 Variables are references

In Python, a variable does not *contain* a value; it *refers* (or points) to an object stored in memory. Python is **dynamically typed**: the same name can refer to objects of different types over time.

````python
x = 1          # x refers to an integer
x = "hello"    # now x refers to a string
x = [1, 2, 3]  # now x refers to a list
````

Because variables are references, two names can refer to the *same* object. If the object can be changed (a list can; we will meet more such types in Module 9), a change made through one name is visible through the other:

````python
x = [1, 2, 3]
y = x          # y refers to the same list as x

x.append(4)    # modify the list through x
print("x =", x)
print("y =", y)
````

Numbers and strings, by contrast, cannot be changed in place. Arithmetic creates a *new* object, and only the name on the left of `=` moves to it:

````python
x = 10
y = x
x = x + 5      # x now refers to a new object, 15; y still refers to 10
print("x =", x)
print("y =", y)
````

### 1.4 Everything is an object

Every value in Python is an **object** with a **type**. The type is attached to the object, not to the variable name. Use `type()` to find it:

````python
x = 4
type(x)
````

````python
x = "hello"
type(x)
````

````python
x = 3.14159
type(x)
````

Objects carry **attributes** (data) and **methods** (functions attached to the object). Both are accessed with a dot. Methods are called with parentheses.

````python
greeting = "hello"
greeting.upper()          # a string method
````

````python
amount = 4.5
amount.is_integer()       # a float method
````

````python
amount.as_integer_ratio() # 4.5 equals 9/2
````

> **Tip.** In Jupyter, type a variable name followed by a dot and press `Tab` to list its attributes and methods. Type a method name followed by `?` (for example `greeting.upper?`) to read its documentation.

### 1.5 Values typed by a user

The `input()` function pauses the program and waits for the user to type something. Whatever the user types is returned as **text** (a string), even if it looks like a number:

```python
answer = input("Enter a temperature: ")   # the user types 33
```

To simulate this, we assign the string directly. Before doing arithmetic, convert the text with `int()` or `float()`:

````python
answer = "33"            # what input() would return if the user typed 33
print(type(answer))

temperature = float(answer)
print(type(temperature), temperature)
````
