## 1 · Preparation

### 1.1 List comprehensions

In Module 7 we built lists with the "empty list and append" pattern:

````python
squares = []
for n in range(12):
    squares.append(n ** 2)
squares
````

A **list comprehension** expresses the same idea in a single line:

````python
[n ** 2 for n in range(12)]
````

Read it almost as English: "the list of `n ** 2` for each `n` in `range(12)`." The general form is

```python
[expression for variable in iterable]
```

A business example: apply a 15% discount to a list of prices.

````python
prices = [19.99, 5.49, 3.75, 12.00]
[round(price * 0.85, 2) for price in prices]
````

### 1.2 Filtering with a condition

Add `if` at the end to keep only the items that satisfy a condition:

````python
[n for n in range(20) if n % 3 > 0]   # numbers that are not multiples of 3
````

````python
orders = [120, 45, 300, 80, 15, 210]
[order for order in orders if order >= 100]   # large orders only
````

The loop equivalent shows where each part of the comprehension comes from:

````python
large = []
for order in orders:          # for order in orders
    if order >= 100:          # if order >= 100
        large.append(order)   # order
large
````

### 1.3 Transforming with a conditional expression

A **conditional expression** chooses between two values in a single line: `value_if_true if condition else value_if_false`.

````python
balance = -250
"overdrawn" if balance < 0 else "ok"
````

Placed at the *front* of a comprehension, it transforms every item (nothing is dropped):

````python
["large" if order >= 100 else "small" for order in orders]
````

> **`if` at the end filters; `if ... else` at the front transforms.**

### 1.4 Multiple loops

A comprehension can contain several `for` clauses, which behave like nested loops:

````python
[(size, color) for size in ["S", "M", "L"] for color in ["red", "blue"]]
````

### 1.5 Set and dictionary comprehensions

Curly braces produce a set comprehension; a `key: value` pair produces a dictionary comprehension:

````python
{n % 3 for n in range(1000)}          # the distinct remainders
````

````python
{product: round(price * 0.85, 2)       # product -> discounted price
 for product, price in [("pen", 2.50), ("notebook", 4.00), ("bag", 30.00)]}
````

### 1.6 Summarizing data

Built-in functions reduce a collection to a single value:

| Function | Returns |
|---|---|
| `len(x)` | the number of items |
| `sum(x)` | the total |
| `min(x)`, `max(x)` | the smallest, largest item |
| `any(x)` | `True` if at least one item is true |
| `all(x)` | `True` if every item is true |

Combined with a comprehension, they answer many analytical questions in one line. When the comprehension is the only argument, the square brackets can be dropped; the result is a **generator expression**, which produces the values one at a time without storing the whole list.

````python
print("total of large orders:", sum(order for order in orders if order >= 100))
print("any order above 250?  ", any(order > 250 for order in orders))
print("all orders positive?  ", all(order > 0 for order in orders))
````

`min()`, `max()`, and `sorted()` accept a `key` function (Module 8), so they can find the item with the largest *something*:

````python
words = ["data", "analytics", "is", "fun"]
max(words, key=len)
````

### 1.7 Map, filter, reduce

Comprehensions implement two of the three classic operations of **functional programming**:

- **map** applies a function to every item (`[f(x) for x in items]`);
- **filter** keeps the items that pass a test (`[x for x in items if test(x)]`);
- **reduce** combines all items into one value.

Built-ins such as `sum()` and `max()` are special cases of reduce. The general version, `functools.reduce`, repeatedly combines a running result with the next item:

````python
from functools import reduce

reduce(lambda total, x: total + x, [1, 2, 3, 4])   # ((1 + 2) + 3) + 4
````

In practice, prefer the specific built-ins (`sum`, `max`, `" ".join`, …) when one exists; they are clearer and faster.

> **Readability first.** Comprehensions are elegant for simple transformations and filters. If you find yourself writing a comprehension that spans several lines with nested conditions, a regular loop is usually easier to read.

## 3 · Workout

````python
msg = """is it possible to star emails from this message for example
if there is an email such as doctor.fun@python.org replace it with stars,
another one is fun.ction@dawg.edu"""
````

**Exercise 3.1.** Replace every email address in `msg` with stars of the same length.

````python
" ".join("*" * len(word) if "@" in word else word for word in msg.split())
````

The conditional expression at the front transforms each word; `" ".join()` reassembles them. (The line breaks are lost because `split()` treats them as spaces.)

**Exercise 3.2.** What is the longest word in `msg`?

**Solution 1.** Find the maximum length, then filter the words that have it:

````python
max_length = max(len(word) for word in msg.split())
[word for word in msg.split() if len(word) == max_length]
````

**Solution 2.** Use `max()` with a `key` function:

````python
max(msg.split(), key=len)
````

**Solution 3.** Use `reduce` to keep the longer of each pair of words:

````python
reduce(lambda w1, w2: w1 if len(w1) >= len(w2) else w2, msg.split())
````

Solution 1 returns *all* the words tied for the maximum length; Solutions 2 and 3 return the first one.

**Exercise 3.3.** What is the longest word that is *not* an email address?

````python
max((word for word in msg.split() if "@" not in word), key=len)
````

The generator expression filters out the emails before `max()` sees them. The extra parentheses are required here because `max()` receives a second argument, `key`.

## Summary

| Pattern | Syntax |
|---|---|
| Transform (map) | `[f(x) for x in items]` |
| Filter | `[x for x in items if condition]` |
| Transform conditionally | `[a if condition else b for x in items]` |
| Nested loops | `[... for x in xs for y in ys]` |
| Set / dictionary | `{x for x in items}` / `{k: v for k, v in pairs}` |
| Reduce | `sum(...)`, `max(...)`, `any(...)`, `all(...)`, `functools.reduce(f, items)` |
