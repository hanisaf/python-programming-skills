## 1 · Preparation

So far our code has run from top to bottom, one line after another. **Control flow** statements change that: they let a program choose between alternatives and repeat steps.

### 1.1 Conditional statements: `if`, `elif`, `else`

An online store charges shipping based on the order total:

```python
order_total = 72.50

if order_total >= 100:
    shipping = 0.00
elif order_total >= 50:
    shipping = 4.99
else:
    shipping = 9.99

print(f"Shipping: ${shipping:.2f}")
```

Python checks the conditions from top to bottom and runs the block under the **first** condition that is `True`; the rest are skipped. `elif` ("else if") can appear any number of times, and both `elif` and `else` are optional. Each `if`, `elif`, and `else` line ends with a colon, and the block below it is indented.

### 1.2 `for` loops

A `for` loop runs a block once for each item in a sequence, such as a list or a string:

```python
for region in ["North", "South", "East", "West"]:
    print("Preparing report for", region)
```

```python
for character in "Python":
    print(character, end=" ")   # end=" " prints a space instead of a newline
```

```python
for i in range(5):
    print(i, end=" ")
```

```python
list(range(5, 10))
```

### 1.3 The accumulator pattern

Many analyses loop over data while updating a running total or count. Initialize the accumulator *before* the loop and update it *inside* the loop:

```python
daily_sales = [1200, 950, 1430, 870, 1610, 2050, 1780]

total = 0
big_days = 0
for sales in daily_sales:
    total += sales
    if sales > 1500:
        big_days += 1

print("Weekly total:", total)
print("Days above $1,500:", big_days)
```

Notice the two levels of indentation: the `if` is inside the loop, and `big_days += 1` is inside the `if`.

### 1.4 `while` loops

A `while` loop repeats as long as its condition stays `True`. Use it when you don't know in advance how many repetitions you need. How many years until an investment of \$1,000 at 7% doubles?

```python
balance = 1000
years = 0
while balance < 2000:
    balance *= 1.07
    years += 1

print(f"{years} years, final balance ${balance:,.2f}")
```

> **Beware of infinite loops.** If the condition never becomes `False`, the loop never ends. In Jupyter, stop a runaway cell with the stop (■) button or *Kernel → Interrupt Kernel*.

### 1.5 `break` and `continue`

- `break` exits the loop immediately.
- `continue` skips the rest of the current iteration and moves to the next one.

```python
for n in range(10):
    if n % 2 == 0:
        continue        # skip even numbers
    print(n, end=" ")
```

Here `break` finds the first day with sales above \$2,000 and stops looking:

```python
for day, sales in enumerate(daily_sales, start=1):
    if sales > 2000:
        print(f"First day above $2,000: day {day} (${sales:,})")
        break
```

`enumerate()` pairs each item with its position. It is the idiomatic way to loop when you need both.

A loop can also have an `else` block, which runs only if the loop finished *without* hitting `break`. This is handy for searches:

```python
for sales in daily_sales:
    if sales < 500:
        print("Found a very slow day:", sales)
        break
else:
    print("No day had sales below $500.")
```

### 1.6 Errors and exceptions

Errors are a normal part of programming. Python reports them by raising an **exception**, whose name tells you what went wrong:

```python
print(undefined_variable)
```

```python
1 / 0
```

```python
1 + "one"
```

```python
float("one")
```

| Exception | Typical cause |
|---|---|
| `NameError` | using a variable that was never defined (often a typo) |
| `ZeroDivisionError` | dividing by zero |
| `TypeError` | an operation applied to the wrong type |
| `ValueError` | the right type with an unacceptable value, such as `float("one")` |
| `IndexError` / `KeyError` | asking for a position or key that does not exist |

When you *expect* that something may fail, for example when converting values typed by users, wrap it in `try`/`except`. If the `try` block raises an exception, Python jumps to the matching `except` block instead of stopping:

```python
# pairs of values as a user might type them
attempts = [("10", "4"), ("10", "0"), ("ten", "2")]

for numerator, denominator in attempts:
    try:
        result = float(numerator) / float(denominator)
        print(f"{numerator} / {denominator} = {result}")
    except ZeroDivisionError:
        print(f"{numerator} / {denominator}: cannot divide by zero")
    except ValueError:
        print(f"{numerator} / {denominator}: only numbers are allowed")
```

Always name the exceptions you expect. A bare `except:` would also hide bugs you did not anticipate, such as a typo in a variable name.

**Exercise 2.2.** Assign 7 to `guess_me` and 1 to `start`. Write a `while` loop that compares `start` with `guess_me`: print `too low` if `start` is lower; if they are equal, print `found it!` and exit the loop; if `start` is greater, print `oops` and exit the loop. Increment `start` at the end of each iteration.

```python
guess_me = 7
start = 1

while True:
    if start < guess_me:
        print("too low")
    elif start == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    start += 1
```

`while True` loops forever unless a `break` ends it, a common pattern when the exit condition is checked in the middle of the loop.

## 3 · Workout

**Exercise 3.1.** Find the numbers hidden in the secret message. A word is a number if every character is a digit or a dash `-`.

```python
secret = """some secret information
the client number is 444-222-5555 and her social
is 345-22-5556"""
```

**Solution 1: a flag variable.** Assume each word is a number until a character proves otherwise.

```python
for word in secret.split():
    is_number = True
    for character in word:
        if character not in "0123456789-":
            is_number = False
            break
    if is_number:
        print(word)
```

**Solution 2: `for`–`else`.** The `else` block runs only when the inner loop did not `break`, that is, when no bad character was found.

```python
for word in secret.split():
    for character in word:
        if character not in "0123456789-":
            break
    else:
        print(word)
```

**Exercise 3.2.** Print the dates of January as `1-1`, `1-2`, …, `1-31`.

```python
for day in range(1, 32):
    print(f"1-{day}", end="  ")
```

**Exercise 3.3.** Extend the loop to print every day of the year, ignoring for now that some months are shorter than 31 days.

**Solution.** A loop inside a loop: for each month, loop over the days. Only the first and last few dates are printed to keep the output short.

```python
dates = []
for month in range(1, 13):
    for day in range(1, 32):
        dates.append(f"{month}-{day}")

print(len(dates), "dates:", dates[:3], "...", dates[-3:])
```

**Exercise 3.4.** Fix the calendar so each month has the right number of days. Use a leap year, such as 2028, in which February has 29 days.

```python
dates = []
for month in range(1, 13):
    if month in [4, 6, 9, 11]:
        days_in_month = 30
    elif month == 2:
        days_in_month = 29
    else:
        days_in_month = 31

    for day in range(1, days_in_month + 1):
        dates.append(f"{month}-{day}")

print(len(dates), "dates:", dates[57:61])
```

The year has 366 dates, and February 29 is followed by March 1.

## Summary

- `if`/`elif`/`else` runs the block of the first true condition.
- `for item in sequence:` repeats a block for each item; `range()` generates integer sequences.
- `while condition:` repeats until the condition is `False`; make sure it eventually is.
- `break` exits a loop; `continue` skips to the next iteration; a loop's `else` runs when no `break` occurred.
- Accumulators (totals, counts, lists) are initialized before a loop and updated inside it.
- `try`/`except SomeError:` handles errors you anticipate. Name the specific exceptions.
