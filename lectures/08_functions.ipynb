## 1 · Preparation

A **function** is a named, reusable block of code. Functions let you write a calculation once and use it many times, give a meaningful name to a step of your analysis, and break a large problem into small pieces that are easy to test.

### 1.1 Calling functions

You have already used many functions: `print()`, `len()`, `type()`, `round()`. A function is called with parentheses that contain its **arguments**. Some arguments are given by position, others by name (**keyword arguments**):

````python
print("Q1", "Q2", "Q3")
````

````python
print("Q1", "Q2", "Q3", sep=" | ")
````

To see which arguments a function accepts, use `help(print)` or type `print?` in a cell.

### 1.2 Defining a function

Use `def`, followed by the function name, the **parameters** in parentheses, and a colon. The indented body runs each time the function is called. `return` sends a result back to the caller.

````python
def net_price(price, discount):
    """Return the price after applying a fractional discount (0.2 means 20% off)."""
    return price * (1 - discount)

net_price(80, 0.25)
````

The string in triple quotes right below the `def` line is a **docstring**. It documents what the function does and appears when someone calls `help(net_price)`. Every function you share with others should have one.

A function can be used anywhere a value can:

````python
total = net_price(80, 0.25) + net_price(120, 0.10)
print(f"Total: ${total:.2f}")
````

**`return` versus `print`.** A function that only *prints* its result gives nothing back to the caller (it returns `None`), so the result cannot be used in further calculations. Prefer `return`; print the result when you call the function.

A function can return several values at once, separated by commas. The caller can unpack them into separate variables:

````python
def summarize(values):
    """Return the minimum, maximum, and average of a list of numbers."""
    return min(values), max(values), sum(values) / len(values)

low, high, average = summarize([1200, 950, 1430, 870, 1610])
print(low, high, average)
````

### 1.3 Default and keyword arguments

Parameters can have **default values**, which are used when the caller leaves them out. Parameters with defaults must come after those without.

````python
def net_price(price, discount=0.0, tax_rate=0.08):
    """Return the final price after a fractional discount, including sales tax."""
    return price * (1 - discount) * (1 + tax_rate)

print(net_price(100))                  # no discount, default tax
print(net_price(100, 0.2))             # 20% discount, default tax
````

Keyword arguments can be given in any order, and they make calls self-explanatory:

````python
net_price(100, tax_rate=0.0, discount=0.5)
````

### 1.4 Flexible arguments: `*args` and `**kwargs`

Occasionally a function should accept any number of arguments. A parameter written `*args` collects extra positional arguments into a tuple, and `**kwargs` collects extra keyword arguments into a dictionary (tuples and dictionaries are covered in Module 9):

````python
def catch_all(*args, **kwargs):
    print("args   =", args)
    print("kwargs =", kwargs)

catch_all(1, 2, 3, a=4, b=5)
````

You will mostly meet this pattern when reading the documentation of libraries such as `pandas` and `seaborn`, whose functions pass extra options along to other functions.

### 1.5 Variables inside functions

Parameters and variables created inside a function are **local**: they exist only while the function runs and are invisible outside it.

````python
def add_tax(amount):
    tax = amount * 0.08     # tax is local to add_tax
    return amount + tax

print(add_tax(50))
````

````python
print(tax)
````

This isolation is a feature: you can use any variable names inside a function without worrying about clashes with the rest of your notebook. Pass everything a function needs as an argument, and get everything back through `return`.

### 1.6 Anonymous functions: `lambda`

A `lambda` expression creates a small function in a single line. It can contain only one expression, whose value is returned automatically:

````python
add = lambda x, y: x + y
add(1, 2)
````

The same function written with `def`:

```python
def add(x, y):
    return x + y
```

`lambda` is most useful for short, throwaway functions passed as arguments to other functions, as shown next.

### 1.7 Functions as arguments

Functions are objects too, so they can be passed to other functions:

````python
def apply_twice(f, x):
    """Apply the function f to x two times."""
    return f(f(x))

def increase(x):
    return x + 1

print(apply_twice(increase, 4))
print(apply_twice(lambda x: x * x, 5))
````

Notice that `increase` is passed *without* parentheses: we hand over the function itself, not the result of calling it.

The most common use is **custom sorting**. `sorted()` takes an optional `key` function, which computes the value to sort by for each item:

````python
numbers = [2, 4, 3, 5, 1, 6, -2, -20]
print(sorted(numbers))
print(sorted(numbers, key=abs))                 # by absolute value
print(sorted(numbers, key=lambda x: -x))        # descending
````

````python
departments = ["Finance", "IT", "Marketing", "HR", "Operations"]
sorted(departments, key=len, reverse=True)      # longest name first
````

## 2 · Warmup

**Exercise 2.1.** Define a function `mask` that takes a string and returns `*` repeated as many times as the string is long. For example, `mask("342-33-4432")` returns `"***********"`.

````python
def mask(text):
    """Return a string of asterisks as long as text."""
    return "*" * len(text)

mask("342-33-4432")
````

**Exercise 2.2.** Define `mask` again using a `lambda` expression.

````python
mask = lambda text: "*" * len(text)
mask("342-33-4432")
````

**Exercise 2.3.** Add an optional parameter `masking_character` whose default value is `"*"`.

````python
def mask(text, masking_character="*"):
    """Return masking_character repeated as many times as text is long."""
    return masking_character * len(text)

print(mask("342-33-4432"))
print(mask("342-33-4432", "#"))
````

**Exercise 2.4.** Define a function `is_number` that returns `True` if a word consists only of digits and dashes, and contains at least one digit.

````python
def is_number(word):
    """Return True if word contains only digits and dashes, with at least one digit."""
    has_digit = False
    for character in word:
        if character not in "0123456789-":
            return False
        if character.isdigit():
            has_digit = True
    return has_digit

print(is_number("333-333"), is_number("phone"), is_number("--"))
````

A `return` inside a loop ends the function immediately, so the loop doesn't need a `break`. The flag `has_digit` rejects words such as `--` that contain dashes only.

A shorter alternative uses string methods:

````python
def is_number(word):
    """Return True if word contains only digits and dashes, with at least one digit."""
    return word.replace("-", "").isdigit()

print(is_number("333-333"), is_number("phone"), is_number("--"))
````

## 3 · Workout

**Exercise 3.1.** Write a function `classify(message)` that hides sensitive numbers in a message by masking them. Use `is_number` and `mask` from the warmup.

````python
msg = """some secret information
the client number is 444-222-5555 and her social
is 345-22-5556 email is precious.client@wealth.org
other than that the weather is sunny and life is good"""
````

````python
def classify(message):
    """Return message with every number replaced by asterisks."""
    for word in message.split():
        if is_number(word):
            message = message.replace(word, mask(word))
    return message

print(classify(msg))
````

**Exercise 3.2.** Numbers are not the only sensitive information: the email address should be hidden too. Make `classify` more general by adding a parameter `is_sensitive`, a *function* that decides whether a word should be masked. Its default should be `is_number`.

````python
def is_email(word):
    """Return True if word looks like an email address."""
    return "@" in word and not word.startswith("@")

def classify(message, is_sensitive=is_number):
    """Return message with every word for which is_sensitive(word) is True masked."""
    for word in message.split():
        if is_sensitive(word):
            message = message.replace(word, mask(word))
    return message

print(classify(msg, is_email))
````

**Exercise 3.3.** Mask both numbers and email addresses.

**Solution 1.** Apply `classify` twice:

````python
print(classify(classify(msg, is_email), is_number))
````

**Solution 2.** Pass a `lambda` that combines the two tests:

````python
print(classify(msg, lambda word: is_email(word) or is_number(word)))
````

Passing behavior as a parameter made `classify` reusable: new kinds of sensitive information (credit card numbers, street addresses, …) can be handled by writing a new test function, without touching `classify` itself.

## Summary

- Define functions with `def name(parameters):`, document them with a docstring, and send results back with `return`.
- Parameters can have default values; callers can pass arguments by position or by keyword.
- Variables created inside a function are local to it.
- `lambda arguments: expression` creates a small anonymous function.
- Functions are objects that can be passed to other functions, e.g. `sorted(items, key=len)`.
