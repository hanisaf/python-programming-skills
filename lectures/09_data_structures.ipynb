## 1 · Preparation

The simple types of Module 5 hold one value each. Real data come in collections: the prices of all products, the grades of all students, the sales of every region. Python has four built-in **compound types** for collections:

| Type | Example | Description |
|---|---|---|
| `list` | `[1, 2, 3]` | ordered, changeable collection |
| `tuple` | `(1, 2, 3)` | ordered, unchangeable collection |
| `dict` | `{"a": 1, "b": 2}` | mapping from unique keys to values |
| `set` | `{1, 2, 3}` | unordered collection of unique values |

Square brackets create a list and parentheses a tuple. Curly braces create a dictionary when they hold `key: value` pairs and a set otherwise; note that `{}` is an empty dictionary, and an empty set is written `set()`.

### 1.1 Lists

A list is an ordered sequence of items, written between square brackets:

````python
prices = [19.99, 5.49, 3.75, 12.00]
print(type(prices), len(prices))
````

Lists are **mutable**: you can add, remove, and change items after the list is created.

````python
prices.append(7.25)         # add one item at the end
prices.extend([2.99, 8.50]) # add several items at the end
prices.insert(0, 99.00)     # insert at position 0
prices
````

````python
removed = prices.pop()      # remove and return the last item
prices.remove(99.00)        # remove the first item equal to 99.00
print(removed, prices)
````

A list can hold items of any type, including other lists:

````python
record = ["Widget", 3, 19.99, ["blue", "red"]]
record[3]
````

**Sorting.** `sorted(items)` returns a new sorted list and leaves the original alone; the method `items.sort()` sorts the list *in place* and returns `None`.

````python
sizes = [5, 2, 8, 1]
print(sorted(sizes), sizes)
sizes.sort(reverse=True)
print(sizes)
````

**Indexing and slicing** work exactly as with strings (Module 6):

```text
 item:        2    3    5    7   11
 index:       0    1    2    3    4
 negative:   -5   -4   -3   -2   -1
```

````python
L = [2, 3, 5, 7, 11]
print(L[0], L[-1])      # first and last items
print(L[1:3])           # items 1 and 2
print(L[::-1])          # reversed copy
````

Unlike strings, lists can be modified through indexing and slicing:

````python
L[0] = 100
L[1:3] = [55, 56]
L
````

### 1.2 Tuples

A tuple is like a list that cannot be changed. Tuples are written with parentheses, or simply with commas:

````python
point = (33.95, -83.37)   # latitude and longitude of Athens, GA
point[0]
````

````python
point[0] = 34.0
````

Use a tuple for a fixed group of values that belong together, such as coordinates or a (name, price) pair. You already used tuples without noticing: a function that returns `a, b` returns a tuple, and `x, y = ...` **unpacks** a tuple into variables.

````python
latitude, longitude = point
print(latitude, longitude)
````

Unpacking also gives a neat way to swap two variables:

````python
a, b = 1, 2
a, b = b, a
print(a, b)
````

### 1.3 Dictionaries

A dictionary maps **keys** to **values**. Instead of looking items up by position, you look them up by key, much like a word in a dictionary. Keys must be unique and of an immutable type (usually strings or numbers).

````python
stock = {"apples": 120, "bananas": 75, "cherries": 0}
stock["bananas"]
````

Assigning to a key adds a new entry or updates an existing one:

````python
stock["dates"] = 40       # add
stock["apples"] -= 20     # update
stock
````

Looking up a missing key raises a `KeyError`. The method `get()` returns a default value instead:

````python
stock["figs"]
````

````python
stock.get("figs", 0)
````

`keys()`, `values()`, and `items()` give access to the contents. Looping over `items()` is the idiomatic way to visit every entry:

````python
print(list(stock.keys()))
print(list(stock.values()))

for fruit, quantity in stock.items():
    if quantity == 0:
        print("Reorder", fruit)
````

**Counting with a dictionary** is one of the most useful patterns in analytics: loop over the data and increase the count of each value you see.

````python
grades = ["A", "B", "B+", "C", "D", "C", "A", "A-", "B+", "A"]

counts = {}
for grade in grades:
    counts[grade] = counts.get(grade, 0) + 1
counts
````

Values can themselves be lists or dictionaries, which lets you represent structured records:

````python
employee = {
    "name": "Ada Lovelace",
    "title": "Analyst",
    "skills": ["Python", "SQL"],
    "address": {"city": "Athens", "state": "GA"},
}
print(employee["skills"][0])
print(employee["address"]["city"])
````

### 1.4 Sets

A set is an unordered collection of **unique** values. Converting a list to a set removes duplicates:

````python
set(grades)
````

Sets support the operations of set algebra, which answer questions such as "which customers bought both products?"

````python
bought_laptop = {"Ann", "Ben", "Cara", "Dev"}
bought_mouse = {"Ben", "Dev", "Eli"}

print(bought_laptop & bought_mouse)   # intersection: bought both
print(bought_laptop | bought_mouse)   # union: bought either
print(bought_laptop - bought_mouse)   # difference: laptop but no mouse
print(bought_laptop ^ bought_mouse)   # symmetric difference: exactly one
````

Each operator has an equivalent method: `intersection()`, `union()`, `difference()`, and `symmetric_difference()`. Checking membership with `in` is much faster on a set than on a long list.

### 1.5 Choosing a data structure

| If you need to… | Use |
|---|---|
| keep items in order and change them | `list` |
| group a fixed number of related values | `tuple` |
| look values up by a name or ID | `dict` |
| keep only unique values, or compare groups | `set` |

## 3 · Workout

News analytics tools scan headlines to find trending topics. Let's build a small one. The text below holds headlines from a news website on September 3, 2019.

````python
headlines = """Hurricane Dorian leaves terrible damage and stranded residents in Bahamas
While residents evacuate, these men and women are flying straight into Hurricane Dorian
Here's what Hurricane Dorian is expected to do as it crawls toward the US
Bahamas' tourism could be devastated for a long time after Hurricane Dorian
Pentagon diverts $3.6 billion in military construction funds to build border wall
These were the passengers and crew members on board the California dive boat
What we know about the California dive boat and its last excursion
Search suspended in the deadly dive-boat fire off California coast
Boris Johnson has just taken a huge gamble over Brexit
Brexit rebels seize Parliament's agenda in major blow to Boris Johnson
The UK could be heading for an early election. Here's what you need to know
What is a no-deal Brexit and what would it mean for Britain?
Kroger asks customers not to openly carry guns in its stores
Walmart ends all handgun ammunition sales and asks customers not to carry guns into stores
Walmart CEO implores Congress to 'do their part' to stop gun violence
This is why Congress remains deadlocked on climate and guns
Elizabeth Warren embraces Jay Inslee's climate change platform
Loud 'boom' reported across Central New York was probably a fireball entering Earth's atmosphere"""
````

**Exercise 3.1.** How many words do the headlines contain? How many *unique* words?

````python
text = headlines.lower()
for mark in ".,?!'":
    text = text.replace(mark, "")

words = text.split()
print("words:       ", len(words))
print("unique words:", len(set(words)))
````

Two cleaning steps come first: converting to lower case ensures that *The* and *the* count as the same word, and removing punctuation ensures that *Bahamas'* and *Bahamas* do too.

**Exercise 3.2.** Count how often each word appears and list the ten most frequent words.

````python
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

ranked = sorted(frequency.items(), key=lambda item: item[1], reverse=True)
ranked[:10]
````

`frequency.items()` yields `(word, count)` tuples. The `key` function tells `sorted()` to order them by the count, `item[1]`.

The most frequent words are small "function words" such as *the* and *to*. They carry little information, so text analysts remove them using a list of **stop words**.

**Exercise 3.3.** Repeat the ranking without stop words.

````python
stop_words = {
    "a", "about", "after", "all", "an", "and", "are", "as", "be", "could", "do",
    "for", "has", "heres", "in", "into", "is", "it", "its", "just", "know",
    "mean", "need", "not", "of", "off", "on", "over", "the", "their", "these",
    "this", "to", "toward", "was", "we", "were", "what", "while", "why", "would",
    "you",
}

ranked_content = []
for word, count in ranked:
    if word not in stop_words:
        ranked_content.append((word, count))

ranked_content[:10]
````

Storing stop words in a **set** makes each `word not in stop_words` check fast, which matters when the text is large. The trending topics of the day emerge: Hurricane Dorian, Brexit, gun sales, and the California dive-boat fire.

**Exercise 3.4.** Python's standard library includes a ready-made counter. Use `collections.Counter` to get the same ranking in two lines.

````python
from collections import Counter

content_words = [word for word in words if word not in stop_words]
Counter(content_words).most_common(10)
````

The first line is a *list comprehension*, a compact way to filter a list. It is the topic of Module 10.

## Summary

- **Lists** (`[...]`) are ordered and mutable: `append`, `extend`, `insert`, `pop`, `remove`, `sort`.
- **Tuples** (`(...)`) are ordered and immutable; unpack them with `a, b = pair`.
- **Dictionaries** (`{key: value}`) look values up by key; loop with `.items()`; count with `d[k] = d.get(k, 0) + 1`.
- **Sets** (`{...}`) hold unique values and support `&`, `|`, `-`, `^`.
- Indexing and slicing work the same way on lists, tuples, and strings.
