## 1 · Preparation

### 1.1 Why Python for business analytics?

Python is a general-purpose programming language that has become the standard tool for data analysis. Three reasons stand out:

- **Readable syntax.** Python code reads almost like English, which makes it easy to learn and easy to review.
- **A rich ecosystem.** Libraries such as `pandas` (data manipulation), `statsmodels` (statistics), and `seaborn` (visualization) cover the entire analytics workflow.
- **Automation and reproducibility.** Unlike point-and-click tools, a script records *every* step of an analysis, so the same result can be produced again — by you, a colleague, or an auditor.

### 1.3 Jupyter notebooks

A notebook is a document made of **cells**:

- **Markdown cells** hold formatted text, like this one.
- **Code cells** hold Python code. When you run a code cell, the result appears right below it.

Behind the notebook is a **kernel**, a running Python process that executes your code and *remembers* every variable you create. The number in `[ ]` next to a code cell records the order in which cells were executed.

| Action | Shortcut |
|---|---|
| Run the current cell and move to the next | `Shift` + `Enter` |
| Enter edit mode / leave edit mode | `Enter` / `Esc` |
| Insert a cell above / below (command mode) | `A` / `B` |
| Change cell to Markdown / Code (command mode) | `M` / `Y` |
| Delete the selected cell (command mode) | `D`, `D` |

Run the cell below. The value of the last line of a cell is displayed automatically.

To display more than one value, or to display text, use the `print()` function:

### 1.4 Reproducible analytics

An analysis is **reproducible** when anyone can run it again from scratch and obtain the same result. Notebooks make this easy, but they also make it easy to break: because the kernel remembers everything, you can run cells out of order and end up with results that no longer match the code on the screen.

Two habits protect you:

1. **Write notebooks that run from top to bottom.** Define something before you use it.
2. **Check your work with *Kernel → Restart & Run All*.** This starts a fresh kernel and runs every cell in order. If the notebook runs without errors, it is reproducible. *Do this before you submit any drill.*

A third habit is to **load data from a stable, shared location** instead of a file that lives only on your laptop. All the datasets in this course are published in the course repository on GitHub, so every notebook reads its data directly from the web:

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/hanisaf/python-programming-skills/main/data/"

salaries = pd.read_csv(DATA_URL + "Cambridge_Budget_Salaries.csv")
salaries.head()
```

**Exercise 3.1.** Take a first look at the Cambridge salary data loaded above.

1. How many rows and columns does the dataset have?
2. What are the names of the columns?
3. What is the largest total salary?

The code below uses a few `pandas` features we have not covered yet. Run it, read it, and see how much of it you can understand from the names alone.

```python
salaries.shape
```

```python
list(salaries.columns)
```

```python
salaries["Total Salary"].max()
```

Even without knowing `pandas`, the code is almost self-explanatory: `shape` gives the number of (rows, columns), `columns` lists the column names, and `max()` finds the largest value of the `Total Salary` column. Readable code is one of Python's biggest strengths.

A "salary" of \$3.5 million should make any analyst suspicious. Let's look at the record behind it:

```python
salaries[salaries["Total Salary"] == salaries["Total Salary"].max()]
```

The record is a *labor settlement* (a lump-sum budget line), not an individual's pay. This is a first lesson in exploration: **always question surprising values** before you report them. Here we would exclude settlements before computing statistics about employee pay.
