Much of the world's business data is text: customer reviews, emails, product descriptions, server logs. Python's `str` type comes with a rich set of tools for working with it.

### 1.1 Creating strings

Single and double quotes are equivalent. Pick the one that avoids a clash with quotes inside the text:

````python
"It's a good day"
````

````python
'He said "good bye"'
````

Triple quotes create a **multi-line string**:

````python
address = """Terry College of Business
600 South Lumpkin Street
Athens, GA 30602"""
print(address)
````

A backslash starts an **escape sequence** that stands for a special character. The most common are `\n` (newline), `\t` (tab), `\"` (a double quote), and `\\` (a backslash).

````python
print("Name:\tAda\nRole:\tAnalyst")
````

### 1.2 Indexing and slicing

A string is a sequence of characters, numbered from zero. Negative indexes count from the end.

```text
 character:   A   n   a   l   y   t   i   c   s
 index:       0   1   2   3   4   5   6   7   8
 negative:   -9  -8  -7  -6  -5  -4  -3  -2  -1
```

````python
word = "Analytics"
print(word[0], word[4], word[-1])
````

A **slice** `word[start:stop]` extracts the characters from `start` up to, but *not including*, `stop`. Either end can be omitted.

````python
print(word[0:4])    # first four characters
print(word[:4])     # same thing
print(word[-3:])    # last three characters
print(word[::-1])   # step of -1: the string reversed
````

Strings are **immutable**: you cannot change a character in place. Every string method returns a *new* string and leaves the original unchanged.

````python
word[0] = "a"
````

### 1.3 Changing case

````python
fox = "tHe qUICk bROWn fOx."
print(fox.upper())
print(fox.lower())
print(fox.title())
print(fox.capitalize())
print(fox.swapcase())
````

Converting to a common case is the standard way to compare text regardless of capitalization: `"Python".lower() == "PYTHON".lower()` is `True`.

### 1.4 Removing unwanted characters

`strip()` removes whitespace (spaces, tabs, newlines) from both ends of a string. `lstrip()` and `rstrip()` work on one end only. Given an argument, they remove those characters instead:

````python
line = "         this is the content         "
line.strip()
````

````python
account = "000000000000435"
account.lstrip("0")
````

### 1.5 Searching

`find()` returns the index of the first occurrence of a substring, or `-1` if it is absent. `index()` does the same but raises an error when the substring is missing. `rfind()` searches from the right, and `count()` counts occurrences.

````python
line = "the quick brown fox jumped over a lazy dog"
print(line.find("fox"))
print(line.find("bear"))
print(line.count("o"))
````

````python
line.index("bear")
````

````python
print(line.startswith("the"))
print(line.endswith("cat"))
print("fox" in line)   # the membership operator from Module 4
````

### 1.6 Replacing

````python
line.replace("brown", "red")
````

### 1.7 Splitting and joining

`split()` breaks a string into a **list** of pieces. With no argument it splits on any whitespace; with an argument it splits on that separator. `splitlines()` splits at line breaks.

````python
line.split()
````

````python
"2026-08-17".split("-")
````

````python
address.splitlines()
````

`join()` is the inverse of `split()`: it glues a list of strings together, inserting the separator between them. Note the unusual syntax: the method is called *on the separator*.

````python
"--".join(["1", "2", "3"])
````

````python
" ".join(["Hello", "analytics", "world"])
````

### 1.8 Inspecting content

Methods starting with `is` test what kind of characters a string contains:

````python
print("2026".isdigit(), "20.26".isdigit())
print("Athens".isalpha(), "Athens GA".isalpha())
````

### 1.9 Aligning text in f-strings

Module 5 introduced format specifications for numbers. Strings have their own: `<` aligns left, `>` right, `^` center, followed by a width.

````python
print(f"|{'left':<10}|{'right':>10}|{'center':^10}|")
print(f"|{'Revenue':<10}|{1234.5:>10,.2f}|")
````

### 1.10 Loading text from the web

The course datasets live in a GitHub repository. Python's built-in `urllib` module can download a file; `read()` returns raw bytes, which `decode("utf-8")` turns into a string.

````python
from urllib.request import urlopen

DATA_URL = "https://raw.githubusercontent.com/hanisaf/python-programming-skills/main/data/"

with urlopen(DATA_URL + "poem.txt") as response:
    poem = response.read().decode("utf-8")

print(poem[:150])
````

The `with` statement opens the connection and closes it automatically when the block ends. With the text in a variable, the tools above apply directly:

````python
lines = poem.splitlines()
words = poem.split()

print("characters:", len(poem))
print("lines:     ", len(lines))
print("words:     ", len(words))
print("first line:", lines[0])
print("'roads' appears", poem.lower().count("roads"), "times")
````

## 3 · Workout

**Exercise 3.1.** Extract the email address from `msg`.

````python
msg = """My name is James Bond and my email address
is james.bond@mi6.co.uk
Do not hesitate to write me. I'm a man for all seasons
"""
````

**Solution 1: search for the boundaries.** Find the `@`, then the nearest whitespace on each side of it.

````python
text = msg.replace("\n", " ")             # treat line breaks as spaces
at = text.find("@")
start = text.rfind(" ", 0, at) + 1        # last space before the @
end = text.find(" ", at)                  # first space after the @
text[start:end]
````

**Solution 2: split around the `@`.** The user name is the last word before the `@`, and the domain is the first word after it.

````python
before, after = msg.split("@")
before.split()[-1] + "@" + after.split()[0]
````

The assignment `before, after = ...` unpacks the two-item list into two variables at once (Module 9 explains why this works).

**Exercise 3.2.** Extract the piece of wisdom that the cow is saying.

````python
the_cow_said = r"""
 ____________________________________
/ "An ounce of prevention is worth a \
\ pound of purge."                   /
 ------------------------------------
        \   ^__^
         \  (oo)\_______
            (__)\       )\/\
                ||----w |
                ||     ||
"""
````

The `r` before the opening quotes makes this a *raw string*: backslashes are kept as ordinary characters instead of starting escape sequences, which is exactly what ASCII art needs.

**Solution.** The wisdom sits between the first and the last double quote. Extract it, then clean it up.

````python
first = the_cow_said.find('"')
last = the_cow_said.rfind('"')
wisdom = the_cow_said[first + 1:last]
wisdom
````

````python
" ".join(wisdom.replace("\\", " ").split())
````

The last step replaces the speech-bubble backslashes with spaces, then uses `split()` and `join()` together, a common idiom that collapses every run of whitespace into a single space.
