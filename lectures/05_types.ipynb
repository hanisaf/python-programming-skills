## 1 · Preparation

Python's simple ("scalar") types hold a single value:

| Type | Example | Description |
|---|---|---|
| `int` | `x = 1` | integers (whole numbers) |
| `float` | `x = 1.0` | floating-point (real) numbers |
| `complex` | `x = 1 + 2j` | complex numbers |
| `bool` | `x = True` | Boolean: `True` or `False` |
| `str` | `x = 'abc'` | strings: text |
| `NoneType` | `x = None` | a special object meaning "no value" |

### 1.1 Integers

Any number without a decimal point is an integer. Python integers have unlimited precision, so they never overflow:

```python
2 ** 200
```

Dividing two integers with `/` always produces a `float`, even when the result is whole. Use `//` to keep an integer:

```python
print(6 / 2, 6 // 2)
```

### 1.2 Floating-point numbers

Floats store fractional numbers. They can be written in decimal or in scientific notation (`e` means "times ten to the power"):

```python
x = 0.000005
y = 5e-6
x == y
```

```python
1.4e6
```

**Floating-point numbers are approximations.** Computers store floats in binary, and most decimal fractions (such as 0.1) have no exact binary representation. The tiny errors usually don't matter, but they can surprise you:

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

Two consequences for business analytics:

1. **Never test floats for exact equality.** Round first, or check that the values are close.
2. **Round only for display**, at the very end of a calculation.

```python
round(0.1 + 0.2, 2) == 0.3
```

```python
import math

math.isclose(0.1 + 0.2, 0.3)
```

### 1.3 Converting between types

The type names double as conversion functions:

```python
print(float(3))        # int -> float
print(int(3.99))       # float -> int: truncates toward zero, does NOT round
print(round(3.99))     # rounding is a separate operation
print(str(42) + "!")   # number -> text
print(int("42") + 8)   # text -> number
```

A conversion that makes no sense raises a `ValueError`:

```python
int("forty-two")
```

### 1.4 Strings

Strings hold text and are written with single or double quotes. Module 6 is devoted to them; here are the essentials.

```python
message = "what do you like?"
response = "spam"

print(len(response))          # number of characters
print(response.upper())       # upper-case copy
print(message.capitalize())   # first letter capitalized
print(message + " " + response)  # concatenation
print(3 * response)           # repetition
```

Individual characters are accessed by **index**, counting from zero. A single character is still a `str`; Python has no separate character type.

```python
message[0]
```

### 1.5 None

`None` is the single value of type `NoneType`. It stands for "nothing" or "missing". For example, a function that does not explicitly return a value returns `None`. The `print()` function displays text but *returns* `None`:

```python
return_value = print("abc")
print(return_value, type(return_value))
```

To check whether a variable is `None`, use the identity operator: `if value is None:`.

### 1.6 Booleans and truthiness

Comparisons produce Booleans, and `bool()` converts other values:

```python
result = 4 < 5
print(result, type(result))
```

Every object has a truth value. The rule is simple: **`None`, zero, and empty things are `False`; everything else is `True`.**

```python
print(bool(2014), bool(0), bool(3.14))
print(bool("abc"), bool(""))
print(bool([1, 2, 3]), bool([]))
print(bool(None))
```

This lets you write conditions such as "if the user typed something" compactly. (The `if` statement is covered in Module 7.)

```python
user_text = ""          # the user pressed Enter without typing
if user_text:
    print("Thank you!")
else:
    print("Please enter a value.")
```

### 1.7 Formatting values with f-strings

An **f-string** is a string with an `f` before the opening quote. Expressions inside curly braces are evaluated and inserted into the text. A **format specification** after a colon controls how the value is displayed:

| Specification | Meaning | Example | Output |
|---|---|---|---|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:,` | thousands separator | `f"{1234567:,}"` | `1,234,567` |
| `:,.2f` | both | `f"{1234.5:,.2f}"` | `1,234.50` |
| `:.1%` | percentage, 1 decimal | `f"{0.256:.1%}"` | `25.6%` |

```python
revenue = 1234567.891
growth = 0.0837
print(f"Revenue: ${revenue:,.2f}")
print(f"Growth:  {growth:.1%}")
```

Formatting changes only how the value is *displayed*; the variable itself keeps full precision.

## 3 · Workout

**Exercise 3.1.** Produce `The grade of Sammy is 88.4%`, keeping one decimal digit.

**Solution 1.** Round the value, then insert it:

```python
rounded_grade = round(grade, 1)
print(f"The grade of {student} is {rounded_grade}%")
```

**Solution 2.** Let the format specification do the rounding:

```python
print(f"The grade of {student} is {grade:.1f}%")
```

Solution 2 is preferable: `grade` keeps its full precision for any later calculation, and rounding happens only at display time.

> **A rounding curiosity.** `round(2.675, 2)` returns `2.67`, not `2.68`. The literal 2.675 is stored in binary as 2.67499999…, which rounds down (see §1.2). This is one more reason to round only at the end.

**Exercise 3.2.** A store sold 1,250 units at \$19.99 each, up from 1,100 units the previous month. Print a one-line report that shows revenue with a thousands separator and two decimals, and unit growth as a percentage with one decimal.

```python
units = 1250
previous_units = 1100
price = 19.99

revenue = units * price
growth = (units - previous_units) / previous_units

print(f"Revenue: ${revenue:,.2f} | Unit growth: {growth:.1%}")
```

## Summary

- `int` and `float` hold numbers; `/` always returns a `float`, `//` keeps integers.
- Floats are approximations: don't compare them with `==`; round only for display.
- Convert with `int()`, `float()`, `str()`, `bool()`. `int()` truncates; `round()` rounds.
- `None` means "no value"; test for it with `is None`.
- `None`, zero, empty strings, and empty containers are falsy; everything else is truthy.
- f-strings (`f"{value:,.2f}"`) are the preferred way to format output.
