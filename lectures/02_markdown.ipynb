## 1 · Preparation

An analysis is only useful if others can follow it. Notebooks let you place explanations right next to the code that produces each result. Those explanations are written in **Markdown**, a lightweight language that turns plain text into formatted text using a handful of symbols such as `#`, `*`, and `-`.

To create a Markdown cell, select a cell and press `M` (or choose *Markdown* from the toolbar). Running the cell renders it; double-clicking a rendered cell shows its source again.

In each section below, the first box shows the **Markdown source** and the text after it shows how it **renders**.

### 1.1 Headings

```markdown
# Heading level 1
## Heading level 2
### Heading level 3
```

Start a line with one to six `#` characters followed by a space. Use headings to give your notebook a clear structure: one level-1 heading for the title, level-2 headings for main sections, and so on.

### 1.6 Tables

```markdown
| Product   | Units | Price  |
|:----------|------:|-------:|
| Laptop    |    12 | \$999  |
| Monitor   |    30 | \$249  |
| Keyboard  |   115 |  \$49  |
```

| Product   | Units | Price  |
|:----------|------:|-------:|
| Laptop    |    12 | \$999  |
| Monitor   |    30 | \$249  |
| Keyboard  |   115 |  \$49  |

Columns are separated by pipes `|`. The second row separates the header from the body and sets the alignment: `:---` left, `---:` right, `:---:` centered. The columns do not need to line up in the source, but aligned source is easier to read.

> **Watch out for dollar signs.** Jupyter treats text between two `$` signs as a math formula. Write `\$` to display a literal dollar sign, as in the table above.

## Summary

| To get | Write |
|---|---|
| Heading | `# Title`, `## Section`, `### Subsection` |
| Bold / italic | `**bold**`, `*italic*` |
| Lists | `1. item` or `- item` (indent sub-items) |
| Link / image | `[text](url)` / `![alt text](url)` |
| Quote | `> quoted text` |
| Code | backticks around `inline code`; three backticks above and below a code block |
| Table | cells separated by pipes, with a row of dashes under the header |
| Math | `$inline$` or `$$displayed$$` |
| Literal dollar sign | `\$` |
