### 1.1 Arithmetic operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `a + b` | Addition | `7 + 2` | `9` |
| `a - b` | Subtraction | `7 - 2` | `5` |
| `a * b` | Multiplication | `7 * 2` | `14` |
| `a / b` | True division | `7 / 2` | `3.5` |
| `a // b` | Floor division | `7 // 2` | `3` |
| `a % b` | Modulus (remainder) | `7 % 2` | `1` |
| `a ** b` | Exponentiation | `7 ** 2` | `49` |
| `-a` | Negation | `-7` | `-7` |

Python follows the usual order of operations: `**` first, then `*`, `/`, `//`, `%`, then `+` and `-`. Use parentheses whenever the order is not obvious.

````python
2 + 4 * 5
````

````python
(2 + 4) * 5
````

Floor division and modulus are a natural pair. For example, a call lasted 135 minutes. How many full hours and leftover minutes is that?

````python
minutes = 135
hours = minutes // 60
leftover = minutes % 60
print(hours, "hours and", leftover, "minutes")
````

Exponentiation appears in many financial formulas. The future value of \$1,000 invested for 10 years at 5% annual interest is $1000 \times (1 + 0.05)^{10}$:

````python
1000 * (1 + 0.05) ** 10
````

### 1.2 Assignment operators

Updating a variable with its own value is so common that Python has a shorthand for every arithmetic operator: `a += b` means `a = a + b`, `a *= b` means `a = a * b`, and so on.

````python
balance = 100
balance += 25    # deposit
balance -= 40    # withdrawal
balance *= 1.02  # 2% interest
balance
````

### 1.3 Comparison operators

Comparisons return a Boolean value, `True` or `False`.

| Operator | Meaning | Operator | Meaning |
|---|---|---|---|
| `a == b` | equal to | `a != b` | not equal to |
| `a < b` | less than | `a > b` | greater than |
| `a <= b` | less than or equal to | `a >= b` | greater than or equal to |

Do not confuse `=` (assignment) with `==` (comparison).

````python
inventory = 25
inventory == 25
````

````python
inventory < 10
````

A number is odd when its remainder after division by 2 is 1:

````python
inventory % 2 == 1
````

Comparisons can be chained, just as in mathematics:

````python
15 < inventory < 30
````

### 1.4 Boolean operators

Combine conditions with `and`, `or`, and `not`:

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `False` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `True` |

````python
credit_score = 710
income = 52000

# approve when the score is good AND income is sufficient
(credit_score >= 680) and (income >= 40000)
````

````python
# flag for review when either condition fails
not (credit_score >= 680) or not (income >= 40000)
````

### 1.5 Identity and membership operators

| Operator | True when |
|---|---|
| `a is b` | `a` and `b` are the *same object* |
| `a is not b` | `a` and `b` are different objects |
| `a in b` | `a` is a member of `b` |
| `a not in b` | `a` is not a member of `b` |

**Identity versus equality.** Two lists can have equal contents and still be different objects:

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)   # same contents
print(a is b)   # same object?
````

````python
c = a
c is a
````

In practice, use `==` to compare values. Reserve `is` for checking against the special value `None` (Module 5), as in `if result is None:`.

**Membership** works with lists and with strings:

````python
2 in [1, 2, 3]
````

````python
"there" in "Hello there"
````

````python
"z" not in "Hello there"
````

### 1.6 Operators on strings and lists

Some operators work on other types, with a meaning adapted to the type. `+` joins (concatenates) two strings or two lists, and `*` repeats them:

````python
"data" + "base"
````

````python
"ha" * 3
````

````python
[1, 2] + [3, 4]
````

**Exercise 3.2.** A tennis club uses the decision tree below to decide whether conditions are right for play.

```text
Outlook?
├── Sunny    → Humidity?  High → don't play  |  Normal → play
├── Overcast → play
└── Rain     → Wind?      Strong → don't play  |  Weak → play
```

Write a single Boolean expression that is `True` exactly when the tree says *play*.

**Solution.** Each path to *play* becomes one condition joined with `and`; the three paths are joined with `or`.

````python
outlook = "Sunny"     # Sunny, Overcast, or Rain
humidity = "Normal"   # High or Normal
wind = "Strong"       # Strong or Weak

play = ((outlook == "Sunny" and humidity == "Normal")
        or outlook == "Overcast"
        or (outlook == "Rain" and wind == "Weak"))
play
````

Wrapping a long expression in parentheses lets it span several lines, which keeps it readable. Change the three variables above and re-run the cell to check every branch of the tree.

## Summary

- Arithmetic: `+ - * / // % **`. Use parentheses to make the order explicit.
- `x += 1` is shorthand for `x = x + 1`; the same works for every arithmetic operator.
- Comparisons (`== != < > <= >=`) return `True` or `False` and can be chained.
- Combine conditions with `and`, `or`, `not`.
- `==` compares values; `is` compares identity (use it with `None`). `in` tests membership.
- `+` and `*` also work on strings and lists, meaning *concatenate* and *repeat*.
